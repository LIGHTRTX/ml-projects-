```python
import numpy as np
import torch
from sklearn.decomposition import PCA
import cv2
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
import tkinter as tk
from tkinter import filedialog

def load_camera_sensitivities(wavelengths):
    def gaussian(x, mu, sigma):
        return np.exp(-((x-mu)**2)/(2*sigma**2))
    cam_r = gaussian(wavelengths, 650, 20)
    cam_g = gaussian(wavelengths, 550, 30)
    cam_b = gaussian(wavelengths, 460, 25)
    cam_r /= cam_r.max(); cam_g /= cam_g.max(); cam_b /= cam_b.max()
    return np.stack([cam_r, cam_g, cam_b], axis=0).astype(np.float32)

def build_pca_basis(spectral_library, n_components=12):
    pca = PCA(n_components=n_components)
    pca.fit(spectral_library)
    mean_spectrum = np.mean(spectral_library, axis=0)
    basis_vectors = pca.components_.T
    return mean_spectrum.astype(np.float32), basis_vectors.astype(np.float32)

def reconstruct_hsi(rgb_img, C, mean_spectrum, basis_vectors, device):
    H, W, _ = rgb_img.shape
    n_components = basis_vectors.shape[1]
    num_bands = C.shape[1]
    C_t = torch.tensor(C, device=device)
    mean_spec_t = torch.tensor(mean_spectrum, device=device)
    basis_t = torch.tensor(basis_vectors, device=device)
    M = torch.matmul(C_t, basis_t)
    M_pinv = torch.pinverse(M)
    img_t = torch.tensor(rgb_img, dtype=torch.float32, device=device)
    img_flat = img_t.view(-1, 3).T
    mean_rgb = (C_t @ mean_spec_t).unsqueeze(1)
    target = img_flat - mean_rgb
    weights = torch.matmul(M_pinv, target)
    spectra = mean_spec_t.unsqueeze(1) + torch.matmul(basis_t, weights)
    spectra = torch.clamp(spectra, 0, 1)
    hsi_cube = spectra.T.view(H, W, num_bands).cpu().numpy()
    return hsi_cube

def select_image_file():
    root = tk.Tk()
    root.withdraw()
    file_path = filedialog.askopenfilename(
        title='Select RGB image file',
        filetypes=[('Image Files', '*.png *.jpg *.jpeg *.tif *.tiff *.bmp'), ('All files', '*.*')]
    )
    return file_path

if __name__ == "__main__":
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    num_bands = 101  # Increased spectral resolution (400-1000nm in 6nm steps)
    wavelengths = np.linspace(400, 1000, num_bands).astype(np.float32)
    num_library_samples = 1500
    library = []
    # Simulate more realistic material classes
    # (You can further expand these or load real spectra for more materials)
    for center, width, height in zip([700, 650, 550, 460, 850, 900],
                                     [80, 50, 60, 70, 35, 25],
                                     [0.1, 0.09, 0.6, 0.2, 0.18, 0.13]):
        base = 0.2 + 0.3*np.random.randn(num_library_samples, num_bands).astype(np.float32)
        base += height * np.exp(-((wavelengths-center)**2)/(2*width**2))
        base = np.clip(base, 0, 1)
        library.append(base)
    spectral_library = np.concatenate(library, axis=0)
    mean_spectrum, basis_vectors = build_pca_basis(spectral_library, n_components=12)
    C = load_camera_sensitivities(wavelengths)

    print("Select your RGB image file:")
    input_image_path = select_image_file()
    if not input_image_path:
        print("No file selected. Exiting.")
        exit(1)
    img = cv2.imread(input_image_path)
    if img is None:
        raise Exception("Could not load image: " + input_image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) / 255.0

    hsi_cube = reconstruct_hsi(img, C, mean_spectrum, basis_vectors, device)
    print(f"HSI cube shape: {hsi_cube.shape}")

    output_hsi_path = input_image_path.rsplit('.',1)[0] + "_uhrgb2hsi_cube.npy"
    np.save(output_hsi_path, hsi_cube)
    print(f"Hyperspectral cube saved as {output_hsi_path}")

    # Visualize 5 bands (first, 1/4, 1/2, 3/4, last)
    band_idxs = [0, num_bands//4, num_bands//2, 3*num_bands//4, num_bands-1]
    plt.figure(figsize=(18,4))
    for i, b in enumerate(band_idxs):
        plt.subplot(1,5,i+1)
        plt.imshow(hsi_cube[:,:,b], cmap='gray')
        plt.title(f'Band {b} ({wavelengths[b]:.0f}nm)')
        plt.axis('off')
    plt.tight_layout()
    plt.show()

    # Interactive band viewer
    fig, ax = plt.subplots(figsize=(5,5))
    plt.subplots_adjust(left=0.1, bottom=0.25)
    band_slider = plt.axes([0.1, 0.1, 0.8, 0.05]) 
    slider = Slider(band_slider, 'Band', 0, num_bands-1, valinit=0, valstep=1)

    img_disp = ax.imshow(hsi_cube[:,:,0], cmap='gray')
    ax.set_title(f'Interactive band viewer ({wavelengths[0]:.0f}nm)')
    ax.axis('off')

    def update(val):
        b = int(slider.val)
        img_disp.set_data(hsi_cube[:,:,b])
        ax.set_title(f'Interactive band viewer ({wavelengths[b]:.0f}nm)')
        fig.canvas.draw_idle()
    slider.on_changed(update)
    plt.show()
```

```python
import numpy as np
import cv2
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

class SpectralFeatureExtractor:
    def __init__(self, n_features=3, n_clusters=4):
        self.n_features = n_features
        self.pca = PCA(n_components=n_features)
        self.scaler = StandardScaler()
        self.kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    
    def fit_transform(self, hsi_cube):
        H, W, B = hsi_cube.shape
        spectra = hsi_cube.reshape(-1, B)
        spectra = self.scaler.fit_transform(spectra)
        features = self.pca.fit_transform(spectra)
        labels = self.kmeans.fit_predict(features)
        return features, labels.reshape(H, W)
    
    def extract_signature(self, hsi_cube, mask=None):
        H, W, B = hsi_cube.shape
        spectra = hsi_cube.reshape(-1, B)
        if mask is not None:
            spectra = spectra[mask.flatten() > 0]
        return np.mean(spectra, axis=0)

if __name__ == "__main__":
    input_image_path = r"C:\Users\LIGHTRQX\Desktop\WhatsApp Image 2025-09-13 at 11.57.12_5b464a6b.jpg"

    img = cv2.imread(input_image_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    hsi_cube = img.astype(np.float32)  # treat RGB as 3-band cube

    extractor = SpectralFeatureExtractor(n_features=3, n_clusters=4)
    features, label_map = extractor.fit_transform(hsi_cube)

    signature = extractor.extract_signature(hsi_cube)
    print("Extracted spectral signature shape:", signature.shape)

    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1)
    plt.imshow(label_map, cmap='tab20')
    plt.title("Cluster Map")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.plot(signature)
    plt.title("Average RGB Signature")
    plt.xlabel("Bands")
    plt.ylabel("Reflectance (normalized)")
    plt.tight_layout()
    plt.show()
```
