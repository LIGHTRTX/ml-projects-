```python
import os
import pickle
import zipfile
import pandas as pd
import scipy.io
import json

BASE_DIRS = [
    r"C:\Users\LIGHTRQX\Desktop\my_babies\DATASET\WESAD",
    r"C:\Users\LIGHTRQX\Desktop\my_babies\DATASET\wearable-device-dataset-from-induced-stress-and-structured-exercise-sessions-1.0.0",
    r"C:\Users\LIGHTRQX\Desktop\my_babies\DATASET\Data"
]

def inspect_csv(file):
    try:
        df = pd.read_csv(file, nrows=5)
        return {"columns": df.columns.tolist(), "shape": df.shape, "sample": df.head().to_dict()}
    except Exception as e:
        return {"error": str(e)}

def inspect_mat(file):
    try:
        data = scipy.io.loadmat(file)
        keys = [k for k in data.keys() if not k.startswith("__")]
        return {"keys": keys}
    except Exception as e:
        return {"error": str(e)}

def inspect_pkl(file):
    try:
        with open(file, "rb") as f:
            obj = pickle.load(f)
        if isinstance(obj, dict):
            return {"type": "dict", "keys": list(obj.keys())}
        elif isinstance(obj, list):
            return {"type": "list", "length": len(obj)}
        else:
            return {"type": str(type(obj))}
    except Exception as e:
        return {"error": str(e)}

def inspect_txt(file):
    try:
        with open(file, "r", encoding="utf-8", errors="ignore") as f:
            lines = [next(f).strip() for _ in range(5)]
        return {"sample_lines": lines}
    except Exception as e:
        return {"error": str(e)}

def inspect_zip(file):
    try:
        with zipfile.ZipFile(file, "r") as z:
            return {"contents": z.namelist()[:10], "total_files": len(z.namelist())}
    except Exception as e:
        return {"error": str(e)}

INSPECTORS = {
    ".csv": inspect_csv,
    ".mat": inspect_mat,
    ".pkl": inspect_pkl,
    ".txt": inspect_txt,
    ".zip": inspect_zip
}

report = []

for base_dir in BASE_DIRS:
    for root, _, files in os.walk(base_dir):
        for file in files:
            path = os.path.join(root, file)
            ext = os.path.splitext(file)[-1].lower()
            inspector = INSPECTORS.get(ext)
            if inspector:
                info = inspector(path)
                report.append({"file": path, "type": ext, "info": info})

# Save full inspection report
pd.DataFrame(report).to_json("dataset_inspection_report.json", indent=2)

print("✅ Inspection complete. Results saved to dataset_inspection_report.json")
```

```python
import os
import glob
import pickle
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

# -------------------------------
# Dataset Loader with Debug Logs
# -------------------------------
class PhysioDataset(Dataset):
    def __init__(self, pkl_paths=None, csv_paths=None, window_size=256, step_size=128):
        self.samples, self.labels = [], []
        loaded_pkl, loaded_csv = 0, 0

        # ---- Load WESAD .pkl files ----
        if pkl_paths:
            for f in pkl_paths:
                try:
                    with open(f, "rb") as handle:
                        data = pickle.load(handle, encoding="latin1")
                    signal_dict = data["signal"]
                    X = np.column_stack([signal_dict[mod] for mod in signal_dict])
                    y = np.array(data["label"])
                    if len(X) < window_size:
                        print(f"[SKIP] {f} (too short: {len(X)} samples)")
                        continue
                    for i in range(0, len(X) - window_size, step_size):
                        self.samples.append(X[i:i+window_size])
                        self.labels.append(np.bincount(y[i:i+window_size]).argmax())
                    loaded_pkl += 1
                    print(f"[OK] Loaded {f} with {len(X)} rows")
                except Exception as e:
                    print(f"[ERROR] Failed to load {f}: {e}")

        # ---- Load Wearable CSV files ----
        if csv_paths:
            for f in csv_paths:
                try:
                    try:
                        df = pd.read_csv(f, encoding="utf-8")
                    except UnicodeDecodeError:
                        df = pd.read_csv(f, encoding="latin1", on_bad_lines="skip")
                except Exception as e:
                    print(f"[ERROR] Cannot read {f}: {e}")
                    continue

                if "label" not in df.columns:
                    print(f"[SKIP] {f} (no label column)")
                    continue

                y = df["label"].values
                X = df.drop(columns=["label"]).select_dtypes(include=[np.number]).values
                if len(X) < window_size:
                    print(f"[SKIP] {f} (too short: {len(X)} samples)")
                    continue

                for i in range(0, len(X) - window_size, step_size):
                    self.samples.append(X[i:i+window_size])
                    self.labels.append(np.bincount(y[i:i+window_size]).argmax())
                loaded_csv += 1
                print(f"[OK] Loaded {f} with {len(X)} rows")

        self.samples = np.array(self.samples, dtype=np.float32)
        self.labels = np.array(self.labels, dtype=np.int64)

        print(f"\n=== SUMMARY ===")
        print(f"WESAD pickle files loaded: {loaded_pkl}")
        print(f"Wearable CSV files loaded: {loaded_csv}")
        print(f"Total samples created: {len(self.samples)}\n")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return torch.tensor(self.samples[idx]), torch.tensor(self.labels[idx])


# -------------------------------
# Simple Fusion Network
# -------------------------------
class FusionNet(nn.Module):
    def __init__(self, input_dim, num_classes=3):
        super(FusionNet, self).__init__()
        self.conv1 = nn.Conv1d(input_dim, 64, kernel_size=5, padding=2)
        self.bn1 = nn.BatchNorm1d(64)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv1d(64, 128, kernel_size=5, padding=2)
        self.bn2 = nn.BatchNorm1d(128)
        self.global_pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Linear(128, num_classes)

    def forward(self, x):
        x = x.permute(0, 2, 1)  # batch, time, features -> batch, features, time
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.global_pool(x).squeeze(-1)
        return self.fc(x)


# -------------------------------
# Paths
# -------------------------------
wesad_paths = glob.glob(r"C:\Users\LIGHTRQX\Desktop\my_babies\DATASET\WESAD\S*\*.pkl")
wearable_csv_paths = glob.glob(r"C:\Users\LIGHTRQX\Desktop\my_babies\DATASET\wearable-device-dataset-from-induced-stress-and-structured-exercise-sessions-1.0.0\**\*.csv", recursive=True)

# -------------------------------
# Dataset + Loader
# -------------------------------
dataset = PhysioDataset(pkl_paths=wesad_paths, csv_paths=wearable_csv_paths)
if len(dataset) == 0:
    print("❌ No samples created. Check logs above.")
else:
    loader = DataLoader(dataset, batch_size=32, shuffle=True)

    # -------------------------------
    # Training Example
    # -------------------------------
    device = "cuda" if torch.cuda.is_available() else "cpu"
    input_dim = dataset.samples.shape[2]
    num_classes = len(np.unique(dataset.labels))

    model = FusionNet(input_dim, num_classes).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    criterion = nn.CrossEntropyLoss()

    for batch_x, batch_y in loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        optimizer.zero_grad()
        out = model(batch_x)
        loss = criterion(out, batch_y)
        loss.backward()
        optimizer.step()
    print("✅ Training step complete.")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, balanced_accuracy_score
from sklearn.utils.class_weight import compute_class_weight
from sklearn.datasets import make_classification
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')


class ImprovedPsychologicalClassifier:
    def __init__(self):
        self.model = None
        self.scaler = StandardScaler()
        self.history = None

    def create_meaningful_sample_data(self, n_samples=2000, n_features=60):
        X, y = make_classification(
            n_samples=n_samples,
            n_features=n_features,
            n_informative=25,
            n_redundant=15,
            n_repeated=5,
            n_classes=4,
            n_clusters_per_class=2,
            weights=[0.35, 0.30, 0.25, 0.10],
            flip_y=0.05,
            random_state=42
        )
        for i in range(n_samples):
            if y[i] == 0:
                X[i, :10] += np.random.normal(0, 0.5, 10)
                X[i, 10:20] *= 0.8
            elif y[i] == 1:
                X[i, :10] += np.random.normal(1.0, 1.0, 10)
                X[i, 20:30] += np.random.normal(0.5, 0.8, 10)
            elif y[i] == 2:
                X[i, 5:15] += np.random.normal(0.7, 1.2, 10)
                X[i, 25:35] *= 1.3
            elif y[i] == 3:
                X[i, :15] += np.random.normal(-0.3, 0.3, 15)
                X[i, 15:30] *= 0.7
        return X, y

    def create_optimized_model(self, input_shape, num_classes=4):
        model = keras.Sequential([
            layers.Dense(256, activation='relu', input_shape=(input_shape,),
                         kernel_regularizer=keras.regularizers.l2(0.001)),
            layers.BatchNormalization(),
            layers.Dropout(0.5),
            layers.Dense(128, activation='relu', kernel_regularizer=keras.regularizers.l2(0.001)),
            layers.BatchNormalization(),
            layers.Dropout(0.4),
            layers.Dense(64, activation='relu'),
            layers.BatchNormalization(),
            layers.Dropout(0.3),
            layers.Dense(32, activation='relu'),
            layers.Dropout(0.2),
            layers.Dense(num_classes, activation='softmax')
        ])
        optimizer = keras.optimizers.Adam(learning_rate=0.0005)
        model.compile(
            optimizer=optimizer,
            loss='sparse_categorical_crossentropy',
            metrics=[tf.keras.metrics.SparseCategoricalAccuracy()]
        )
        return model

    def improved_imbalance_handling(self, X, y, use_smote=True):
        class_weights = compute_class_weight('balanced', classes=np.unique(y), y=y)
        class_weight_dict = {i: weight for i, weight in enumerate(class_weights)}
        if use_smote and np.min(np.bincount(y)) >= 5:
            k_neighbors = min(3, np.min(np.bincount(y)) - 1)
            smote = SMOTE(random_state=42, k_neighbors=k_neighbors)
            X_res, y_res = smote.fit_resample(X, y)
            return X_res, y_res, None  # Do not use class_weight after SMOTE
        return X, y, class_weight_dict

    def train_with_cross_validation(self, features, labels, n_splits=5):
        skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        fold_accuracies = []
        fold_histories = []
        best_model = None
        best_accuracy = 0

        for fold, (train_idx, val_idx) in enumerate(skf.split(features, labels)):
            X_train, X_val = features[train_idx], features[val_idx]
            y_train, y_val = labels[train_idx], labels[val_idx]
            scaler = StandardScaler()
            X_train_scaled = scaler.fit_transform(X_train)
            X_val_scaled = scaler.transform(X_val)
            X_train_balanced, y_train_balanced, _ = self.improved_imbalance_handling(X_train_scaled, y_train)

            model = self.create_optimized_model(X_train_balanced.shape[1])

            callbacks = [
                keras.callbacks.EarlyStopping(monitor='val_sparse_categorical_accuracy',
                                              patience=15,
                                              restore_best_weights=True),
                keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=8, min_lr=1e-7)
            ]

            history = model.fit(X_train_balanced, y_train_balanced,
                                validation_data=(X_val_scaled, y_val),
                                epochs=100, batch_size=32,
                                verbose=0)
            val_pred = np.argmax(model.predict(X_val_scaled), axis=1)
            accuracy = accuracy_score(y_val, val_pred)
            fold_accuracies.append(accuracy)
            fold_histories.append(history)
            if accuracy > best_accuracy:
                best_accuracy = accuracy
                best_model = model

        self.model = best_model
        self.history = fold_histories[np.argmax(fold_accuracies)]
        return np.mean(fold_accuracies)

    def comprehensive_evaluation(self, features, labels, test_size=0.2):
        X_train, X_test, y_train, y_test = train_test_split(
            features, labels, test_size=test_size, random_state=42, stratify=labels
        )
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_test_scaled = self.scaler.transform(X_test)
        X_train_balanced, y_train_balanced, _ = self.improved_imbalance_handling(X_train_scaled, y_train)
        self.model = self.create_optimized_model(X_train_balanced.shape[1])

        callbacks = [
            keras.callbacks.EarlyStopping(monitor='val_sparse_categorical_accuracy',
                                          patience=20,
                                          restore_best_weights=True)
        ]

        self.history = self.model.fit(
            X_train_balanced, y_train_balanced,
            validation_data=(X_test_scaled, y_test),
            epochs=100, batch_size=32,
            verbose=1
        )

        y_pred = np.argmax(self.model.predict(X_test_scaled), axis=1)
        acc = accuracy_score(y_test, y_pred)
        bal_acc = balanced_accuracy_score(y_test, y_pred)
        print(f"\nTest Accuracy: {acc:.4f}")
        print(f"Balanced Accuracy: {bal_acc:.4f}")
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred, target_names=['Neutral', 'Stress', 'Anxiety/Fear', 'Meditation']))
        self.plot_confusion_matrix(y_test, y_pred)
        return acc

    def plot_confusion_matrix(self, y_true, y_pred):
        cm = confusion_matrix(y_true, y_pred)
        cm_percent = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax1,
                    xticklabels=['Neutral', 'Stress', 'Anxiety/Fear', 'Meditation'],
                    yticklabels=['Neutral', 'Stress', 'Anxiety/Fear', 'Meditation'])
        ax1.set_title('Counts')
        sns.heatmap(cm_percent, annot=True, fmt='.2%', cmap='Blues', ax=ax2,
                    xticklabels=['Neutral', 'Stress', 'Anxiety/Fear', 'Meditation'],
                    yticklabels=['Neutral', 'Stress', 'Anxiety/Fear', 'Meditation'])
        ax2.set_title('Percentages')
        plt.show()


def main():
    classifier = ImprovedPsychologicalClassifier()
    X, y = classifier.create_meaningful_sample_data()
    print(f"Dataset: {X.shape[0]} samples, {X.shape[1]} features")
    print(f"Class distribution: {np.bincount(y)}")
    cv_acc = classifier.train_with_cross_validation(X, y)
    print(f"\nCross-Validation Accuracy: {cv_acc:.4f}")
    final_acc = classifier.comprehensive_evaluation(X, y)
    print(f"\nFinal Model Accuracy: {final_acc:.4f}")


if __name__ == "__main__":
    main()
```
